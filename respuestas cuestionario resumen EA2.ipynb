```python
#problema 14
#Servicio A (2 1)(x) = (10)
#Servicio B (3 -2)(y) = (8)


#
F = np.array([[2, 1],
              [3,-2]])  #Tengo que tener cuidado era un numero negativo
B = np.array([[10],
              [8]])


F_inv = np.linalg.inv(F)
C=np.dot(F_inv,B)
print(C)

```
