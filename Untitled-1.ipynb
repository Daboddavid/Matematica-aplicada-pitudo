```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.optimize import fsolve



def s(x):
    return -1.6*x**2+1.3*x+141.3

def r(x):
    return 1.4*x**2 - 5.6*x + 12.6


def f(x):
    return r(x) - s(x)

#Valor/es inicial/es de la aproximación
xo = np.linspace(-5, 10, 4)
solucion = fsolve(f, xo)
print(f"es {solucion}")



x = np.arange(-20, 12, 1)
plt.plot(x, s(x), label = 's(x) auto')
plt.plot(x, r(x), label = 'r(x) moto')
plt.title('Nombre del gráfico')
plt.ylabel('Nombre eje y')
plt.xlabel('Nombre eje x')
plt.legend()
plt.grid()
plt.show()



```
